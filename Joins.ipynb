# Joins and Concatenations
This notebooks is dedicated to the introduction of joins and concatenations in Pandas. These operations allow you to merge or append DataFrames based on specific keys or conditions.

**Remark**: there is also `append()` method in pandas, but we will not cover it here as it is rarely used in practice.


## Joins
Joins (also called **merge**) allow for combining data from multiple DataFrames.

There are different types of joins depending on how you want to align the data.
Types of Joins in Pandas:

1. Inner Join `pd.merge(how='inner)`: Returns only rows with matching keys in both DataFrames (like an SQL INNER JOIN).
2. Left Join `pd.merge(how='left')`: Returns all rows from the left DataFrame and the matching rows from the right DataFrame.
3. Right Join `pd.merge(how='right')`: Returns all rows from the right DataFrame and the matching rows from the left DataFrame.
4. Outer Join `pd.merge(how='outer')`: Returns all rows from both DataFrames, filling missing values with NaN.

#### Example
Let us now perform simple merges using dummy datasets.

**Hint**: Check [pd.merge()](https://pandas.pydata.org/docs/reference/api/pandas.merge.html) before.

````python
# Sample DataFrames
df_1 = pd.DataFrame({'ID': [1, 2, 3], 'Name': ['Alice', 'Bob', 'Charlie']})
df_2 = pd.DataFrame({'ID': [2, 3, 4], 'Department': ['HR', 'IT', 'Finance']})

# Inner Join
df_inner = pd.merge(df_1, df_2, on='ID', how='inner'
                    )
print(df_inner)

# Left Join
df_left = pd.merge(df_1, df_2, on='ID', how='left'
                    )
print(df_left)

# Outer Join
df_outer = pd.merge(df_1, df_2, on='ID', how='outer'
                    )
print(df_outer)
````

#### Exercise 2
Imagine you are managing data from a car manufacturing company. One DataFrame contains information about the car models and their prices, and another contains details about the sales of these models in various regions. You need to merge the two DataFrames to analyze the relationship between the cars' prices and their sales. Perform the following tasks:
* Merge the two datasets where the car model matches
* Merge the two datasets where the car model the year of manufacturing match

````python
# DataFrame 1: Car details
df_cars = pd.DataFrame({
    'Car': ['Audi', 'BMW', 'Mercedes', 'Toyota'],
    'Price': [35000, 45000, 50000, 25000],
    'Year': [2015, 2017, 2018, 2020]
})

# DataFrame 2: Sales data by region
df_sales = pd.DataFrame({
    'Car': ['Audi', 'Audi', 'BMW', 'BMW', 'Toyota'],
    'Year': [2015, 2017, 2017, 2020, 2020],
    'Region': ['Europe', 'Asia', 'North America', 'Asia', 'Europe'],
    'Sales': [1500, 1700, 1200, 1400, 2000]
})

# Perform a join on car model
df_merged_1 = pd.merge(df_cars, df_sales, on='Car', how='inner'
                     )
print(df_merged_1)

# Perform a join on car model and year of production
df_merged_2 = pd.merge(df_cars, df_sales, on=['Car','Year'], how='inner'
                     )
print(df_merged_2)
````

**Question:** How would you perform the merge if the column names were different? See below.

````python
# DataFrame 1: Car details
df_cars = pd.DataFrame({
    'Car': ['Audi', 'BMW', 'Mercedes', 'Toyota'],
    'Price': [35000, 45000, 50000, 25000],
    'Year': [2015, 2017, 2018, 2020]
})

# DataFrame 2: Sales data by region
df_sales = pd.DataFrame({
    'Auto': ['Audi', 'Audi', 'BMW', 'BMW', 'Toyota'],
    'Year': [2015, 2017, 2017, 2020, 2020],
    'Region': ['Europe', 'Asia', 'North America', 'Asia', 'Europe'],
    'Sales': [1500, 1700, 1200, 1400, 2000]
})

# Perform a join on car model
df_merged_1 = pd.merge(df_cars, df_sales, left_on='Car',right_on='Auto'
                     )
print(df_merged_1)

# Perform a join on car model and year of production
df_merged_2 = pd.merge(df_cars, df_sales, left_on=['Car','Year'], right_on=['Auto','Year']
                     )
print(df_merged_2)
````

## Concatenations
Concatenation is the process of appending one DataFrame to another, either vertically (stacking rows) or horizontally (adding columns). You can concatenate along either axis (*axis=0* or *axis=1*):
* `pd.concat([df_1, ..., df_n], axis=0)`: This function can concatenate DataFrames along rows.
* `pd.concat([df_1, ..., df_n], axis=1)`: This function can concatenate DataFrames along columns.

#### Example
Let us perform simple concatenations on the following dummy DataFrames.
